```python
import scanpy as sc
import squidpy as sq
import matplotlib.pyplot as plt

# Load your combined Visium data
adata = sc.read_h5ad("./GSE171351_combined_visium_with_images.h5ad")

# Inspect the object
print(adata)
print(adata.obs.head())
print(adata.obsm.keys())



import os
from skimage.io import imread
from skimage.transform import resize
from PIL import Image
Image.MAX_IMAGE_PIXELS = None  # disable decompression bomb check

# Define folder where your images are located
img_folder = "./spatial_bladder_4"

# Map library IDs to image filenames
image_files = {
    "A1": "GSM5224027_A1.jpg",
    "B1": "GSM5224028_B1.jpg",
    "C1": "GSM5224029_C1.jpg",
    "D1": "GSM5224030_D1.jpg"
}

# Desired maximum image dimension
max_dim = 5000  # e.g., 5000 pixels for width or height

for lib_id, filename in image_files.items():
    img_path = os.path.join(img_folder, filename)
    if not os.path.exists(img_path):
        print(f"Warning: {img_path} not found!")
        continue

    # Load the image
    img = imread(img_path)
    orig_height, orig_width = img.shape[:2]

    # Calculate scaling factor to fit within max_dim
    scale = max_dim / max(orig_height, orig_width)
    if scale < 1.0:  # only resize if the image is larger than max_dim
        new_height = int(orig_height * scale)
        new_width = int(orig_width * scale)
        img_resized = resize(img, (new_height, new_width), anti_aliasing=True)
    else:
        img_resized = img
        scale = 1.0  # no scaling needed

    # Attach resized image to AnnData
    if 'spatial' not in adata.uns:
        adata.uns['spatial'] = {}
    if lib_id not in adata.uns['spatial']:
        adata.uns['spatial'][lib_id] = {}

    adata.uns['spatial'][lib_id]['images'] = {'hires': img_resized}

    # Rescale spatial coordinates to match resized image
    if lib_id in adata.obsm['spatial']:
        adata.obsm['spatial'][lib_id][:, 0] *= scale  # x coordinates
        adata.obsm['spatial'][lib_id][:, 1] *= scale  # y coordinates

# Save updated AnnData
adata.write_h5ad("GSE171351_combined_visium_with_images.h5ad")
print("Images attached, coordinates rescaled, and AnnData saved.")
```
